### Pre-requisites : Loading data

```python
%load_ext autoreload
%autoreload 2
%matplotlib qt

# Generic imports

# espm imports
import espm.datasets as ds
from espm.estimators import SmoothNMF

# Generate the data
# Here `seeds_range` is the number of different samples to generate
ds.generate_built_in_datasets(seeds_range=1)

# Load the data
spim = ds.load_particules(sample=0)
# We need to change the data to floats for the decomposition to run
spim.change_dtype("float64")
spim.build_G()
```

### 1.1. cross sections utils

```python
from espm.cross_sections_utils import modify_cross_sections

modify_cross_sections(
    100,
    input_type="scaling_factors",
    lines_scaling_factors={"Ca_Ka": 0.9, "W_La1": 1.1},
)
```

### 1.2. eds_spim

#### 1.2.1. Fixed H

```python
roi = spim.define_ROI(xray_lines=False)
```

```python
fH_0 = spim.generate_part_fixed_H_matrix(rois=[roi])
fH_1 = spim.generate_part_fixed_H_matrix(rois=None)
fH_2 = spim.generate_part_fixed_H_matrix(rois=[])

fH_dict = {"0": fH_0, "1": fH_1, "2": fH_2}

fH = spim.set_fixed_H(fH_dict)
```

```python
snmf = SmoothNMF(
    n_components=3,
    fixed_H=fH,
    hspy_comp=True,
    simplex_H=True,
    simplex_W=False,
    G=spim.model,
)
spim.decomposition(algorithm=snmf)
```
